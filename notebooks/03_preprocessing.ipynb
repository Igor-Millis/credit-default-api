```python
# ============================================================
# 1. IMPORTAÇÃO DAS BIBLIOTECAS
# ============================================================

import pandas as pd

from sklearn.model_selection import train_test_split

from sklearn.compose import ColumnTransformer

from sklearn.pipeline import Pipeline

from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder
)
```

```python
# ============================================================
# 2. LEITURA DOS DADOS
# ============================================================

df = pd.read_excel(
    "C:/Users/T-Gamer/Desktop/igor/credit-default-api/data/raw/default_of_credit_card_clients.xls",
    header=1)

df.head()
```

```python
# ============================================================
# 3. SEPARAÇÃO ENTRE VARIÁVEIS PREDITORAS E VARIÁVEL RESPOSTA
# ============================================================

X = df.drop(columns=["default payment next month"])

y = df["default payment next month"]
```

```python
# ============================================================
# 4. REMOÇÃO DA COLUNA IDENTIFICADORA
# ============================================================

X = X.drop(columns=["ID"])
```

```python
print(f"Número de observações: {X.shape[0]}")
print(f"Número de variáveis preditoras: {X.shape[1]}")
print(f"Variável resposta: {y.name}")
```

```python
numeric_features = [
    "LIMIT_BAL",
    "AGE",
    "BILL_AMT1", "BILL_AMT2", "BILL_AMT3",
    "BILL_AMT4", "BILL_AMT5", "BILL_AMT6",
    "PAY_AMT1", "PAY_AMT2", "PAY_AMT3",
    "PAY_AMT4", "PAY_AMT5", "PAY_AMT6"
]

categorical_features = [
    "SEX",
    "EDUCATION",
    "MARRIAGE"
]

ordinal_features = [
    "PAY_0",
    "PAY_2",
    "PAY_3",
    "PAY_4",
    "PAY_5",
    "PAY_6"
]

all_features = (
    numeric_features +
    categorical_features +
    ordinal_features
)

print(f"Variáveis classificadas: {len(all_features)}")
print(f"Variáveis em X: {X.shape[1]}")
```

```python
# ============================================================
# DIVISÃO TREINO / TESTE
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print(X_train.shape)
print(X_test.shape)

print(y_train.value_counts(normalize=True))
print(y_test.value_counts(normalize=True))
```

```python
numeric_transformer = Pipeline(
    steps=[
        ("scaler", StandardScaler())
    ]
)

categorical_transformer = Pipeline(
    steps=[
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                drop="first"
            )
        )
    ]
)

```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features),
        ("ord", "passthrough", ordinal_features)
    ]
)
```
