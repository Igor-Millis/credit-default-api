```python
# ============================================================
# CREDIT DEFAULT API
# 05_modeling.py
# Modelo 1 - Regressão Logística
# ============================================================


# ============================================================
# 1. IMPORTAÇÃO DAS BIBLIOTECAS
# ============================================================

import pandas as pd

import joblib

from sklearn.model_selection import train_test_split

from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)


# ============================================================
# 2. LEITURA DOS DADOS
# ============================================================

df = pd.read_excel(
    "C:/Users/T-Gamer/Desktop/igor/credit-default-api/data/raw/default_of_credit_card_clients.xls",
    header=1)


# ============================================================
# 3. SEPARAÇÃO X E Y
# ============================================================

X = df.drop(
    columns=[
        "ID",
        "default payment next month"
    ]
)


y = df[
    "default payment next month"
]


# ============================================================
# 4. DIVISÃO TREINO E TESTE
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


# ============================================================
# 5. DEFINIÇÃO DAS VARIÁVEIS
# ============================================================


numeric_features = [
    "LIMIT_BAL",
    "AGE",

    "BILL_AMT1",
    "BILL_AMT2",
    "BILL_AMT3",
    "BILL_AMT4",
    "BILL_AMT5",
    "BILL_AMT6",

    "PAY_AMT1",
    "PAY_AMT2",
    "PAY_AMT3",
    "PAY_AMT4",
    "PAY_AMT5",
    "PAY_AMT6"
]


categorical_features = [
    "SEX",
    "EDUCATION",
    "MARRIAGE"
]


ordinal_features = [
    "PAY_0",
    "PAY_2",
    "PAY_3",
    "PAY_4",
    "PAY_5",
    "PAY_6"
]


# ============================================================
# 6. PRÉ-PROCESSAMENTO
# ============================================================

from sklearn.compose import ColumnTransformer

from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder
)


numeric_transformer = Pipeline(
    steps=[
        (
            "scaler",
            StandardScaler()
        )
    ]
)


categorical_transformer = Pipeline(
    steps=[
        (
            "onehot",
            OneHotEncoder(
                drop="first",
                handle_unknown="ignore"
            )
        )
    ]
)


preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_transformer,
            numeric_features
        ),

        (
            "cat",
            categorical_transformer,
            categorical_features
        ),

        (
            "ord",
            "passthrough",
            ordinal_features
        )
    ]
)



# ============================================================
# 7. CRIAÇÃO DO PIPELINE FINAL
# ============================================================


logistic_pipeline = Pipeline(
    steps=[

        (
            "preprocessor",
            preprocessor
        ),

        (
            "model",
            LogisticRegression(
                random_state=42,
                max_iter=1000
            )
        )
    ]
)



# ============================================================
# 8. TREINAMENTO
# ============================================================


logistic_pipeline.fit(
    X_train,
    y_train
)



# ============================================================
# 9. PREDIÇÕES
# ============================================================


y_pred = logistic_pipeline.predict(
    X_test
)


y_prob = logistic_pipeline.predict_proba(
    X_test
)[:,1]



# ============================================================
# 10. AVALIAÇÃO
# ============================================================


accuracy = accuracy_score(
    y_test,
    y_pred
)


precision = precision_score(
    y_test,
    y_pred
)


recall = recall_score(
    y_test,
    y_pred
)


f1 = f1_score(
    y_test,
    y_pred
)


roc_auc = roc_auc_score(
    y_test,
    y_prob
)



print("===== RESULTADOS REGRESSÃO LOGÍSTICA =====")

print(
    f"Accuracy : {accuracy:.4f}"
)

print(
    f"Precision: {precision:.4f}"
)

print(
    f"Recall   : {recall:.4f}"
)

print(
    f"F1 Score : {f1:.4f}"
)

print(
    f"ROC AUC  : {roc_auc:.4f}"
)



# ============================================================
# 11. RELATÓRIO DE CLASSIFICAÇÃO
# ============================================================


print(
    classification_report(
        y_test,
        y_pred
    )
)



# ============================================================
# 12. MATRIZ DE CONFUSÃO
# ============================================================


cm = confusion_matrix(
    y_test,
    y_pred
)


print(cm)



# ============================================================
# 13. SALVAR MODELO
# ============================================================


joblib.dump(
    logistic_pipeline,
    "../models/logistic_regression_pipeline.pkl"
)


print(
    "Modelo salvo com sucesso!"
)
```
